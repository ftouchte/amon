```python
%jsroot on

import ROOT
import array
import csv

# -------- lecture du CSV exporté depuis le hipo --------
input_csv = "./files/graph_export.csv"

x_list, y_list, ex_list, ey_list = [], [], [], []
with open(input_csv, newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        x_list.append(float(row["x"]))
        y_list.append(float(row["y"]))
        ex_list.append(float(row["ex"]))
        ey_list.append(float(row["ey"]))

x  = array.array('d', x_list)
y  = array.array('d', y_list)
ex = array.array('d', ex_list)
ey = array.array('d', ey_list)

# -------- canvas --------
canvas = ROOT.TCanvas("canvas", "Mon graphe", 1500, 700)
canvas.SetRightMargin(0.05)

graph = ROOT.TGraphErrors(len(x), x, y, ex, ey)
graph.SetTitle("Mon graphe;Run;VZ upstream (cm)")
graph.SetMarkerStyle(20)
graph.SetMarkerColor(ROOT.kBlue)
graph.Draw("AP")

canvas.Draw()
canvas.SaveAs("plot.pdf")
```
